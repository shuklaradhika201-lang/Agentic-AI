```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:qwen/qwen3.8-27b")

response=model.invoke("why do people talk")
response
```

text prompts- they are strings. these are ideal for straightforward genration taska where you dont need to retain conversation history

use text prompt when - you have single standalone request , you dont need conversation history , you want minimal code complexity 

MESSAGEPROMPT- you can pass the list of message to the model by providing a list of message as a object

SYSTEMMESSAGE- these are the list of instruction of how model should behave 

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages=[
    SystemMessage("you are a poetry expert"),
    HumanMessage("write a poem about artifical inteligence")
]

response=model.invoke(messages)
response.content
```

```python
system_msg=SystemMessage("you are a helpful coding assistant")
messages=[
    system_msg,
    HumanMessage("how do i create a rest api")
]

response=model.invoke(messages)
print(response.content)
```

```python
## detailed info to the llm thorugh system message
from langchain.messages import SystemMessage,HumanMessage

system_msg=SystemMessage("""you are a senior python developer
 with expertise in web frameworks , always provide code examples and explain your reasoning. be concise but through in your explanantion""")

messages=[
    system_msg,
    HumanMessage("how do i create a REST API")
]

response=model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import AIMessage, SystemMessage , HumanMessage

ai_msg=AIMessage("i'd be happy to help you with that question")

messages=[
    SystemMessage("you are a helpful assistant"),
    HumanMessage("can you help me"),
    ai_msg,
    HumanMessage("great ! what's 2+2 ?")
]

response=model.invoke(messages)
response.content
```
