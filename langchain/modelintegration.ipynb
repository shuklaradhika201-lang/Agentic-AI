```python
from langchain.chat_models import init_chat_model
model = init_chat_model("groq:qwen/qwen3.8-27b")

for chunk in model.stream("write an easyy of 200 words on artifical intelligence"):
    print(chunk.text , end="" )
```

```python
for chunk in model.stream("why do parrots have colorful feathers"):
    print(chunk.text , end="")
```

```python
# basically what is stream 
# we can just print the output while it is geenrating one word sfter word the genrator dont have to wait for complete answer to print
#in the case of invoke when the entire response is generated then only it will print but in streaming as it is genrating we can see the response
```

```python
#BATCH
from langchain.chat_models import init_chat_model
model = init_chat_model("groq:qwen/qwen3.8-27b")
responses = model.batch(
    [
        "why do parrots have colorful feathers",
        "how do airplanes fly",
        "what is quantum computing",
    ],
    config={
        "max_concurrency": 5
        #it tolds about the limit of parallel calls that can be made 
        #here we can ask max 5 questions parallelly
    }
)

for response in responses:
    print(response.content)
```
